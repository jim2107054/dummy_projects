This notebook acts as a metric validation harness to score generated responses locally.

### 1. Score Validation predictions

```python
import os
import sys
import pandas as pd

sys.path.append(os.path.abspath('src'))
import metric_utils

pred_path = "/kaggle/working/val_predictions.csv"
val_path = "/kaggle/working/sft_val.csv"

if os.path.exists(pred_path):
    print(f"Scoring {pred_path} against {val_path}...")
    score = metric_utils.score_predictions_csv(pred_path, val_path)
    print(f"\nLocal Composite Score: {score:.4f}")
else:
    print("No predictions found. Please run '03_inference_and_submit.ipynb' in validation mode first.")
```

### 2. Metric Harness Inline Unit Tests
Let's sanity check metric_utils.py on near-identical and unrelated sentence pairs.

```python
# Near-identical sentences
pred_1 = "হ্যালো, আমি মনে করি আপনার নিয়মিত ব্যায়াম করা এবং প্রচুর পানি পান করা উচিত।"
ref_1  = "হ্যালো, আমি বিশ্বাস করি আপনার প্রতিদিন ব্যায়াম করা এবং পর্যাপ্ত পানি খাওয়া উচিত।"

# Unrelated sentences
pred_2 = "হ্যালো, আমি মনে করি আপনার নিয়মিত ব্যায়am করা উচিত।"
ref_2  = "অ্যান্টিবায়োটিক গ্রহণের পূর্বে অবশ্যই একজন অভিজ্ঞ চিকিৎসকের পরামর্শ নিন।"

print("=== Running Metric Unit Tests ===")
print(f"\nPair 1 (Similar):")
print("Pred:", pred_1)
print("Ref: ", ref_1)
score_1, df_1 = metric_utils.composite_score([pred_1], [ref_1])
print(f"Composite Score: {score_1:.4f} | Token F1: {df_1.iloc[0]['token_level_f1']:.4f} | ROUGE-L: {df_1.iloc[0]['rouge_l_f1']:.4f}")

print(f"\nPair 2 (Unrelated):")
print("Pred:", pred_2)
print("Ref: ", ref_2)
score_2, df_2 = metric_utils.composite_score([pred_2], [ref_2])
print(f"Composite Score: {score_2:.4f} | Token F1: {df_2.iloc[0]['token_level_f1']:.4f} | ROUGE-L: {df_2.iloc[0]['rouge_l_f1']:.4f}")

assert score_1 > score_2, "Sanity check failed: Similar sentences must score higher than unrelated ones!"
print("\n[PASS] Metric harness unit tests passed successfully!")
```
